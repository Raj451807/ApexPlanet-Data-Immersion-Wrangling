```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# load Excel File
df = pd.read_excel(r"/content/ApexPlanet_DataAnalytics_Dataset.xlsx")

# Display First 5 Rows
print(df.head())

# Dataset Information
print(df.info())

# Dataset dimentions
print(df.shape)
```

```python
# Now Data Quality Assessment

#Check Missing Values
print(df.isnull().sum())

missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
# Check complete duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Check duplicate Order IDs
print("Duplicate Order IDs:",
      df["Order_ID"].duplicated().sum())

# Display duplicated Order IDs
print(df[df["Order_ID"].duplicated(keep=False)])

```

```python
print(df.dtypes)
```

```python
# Checking The Outliers

numeric_columns = ["Age","Quantity","Unit_Price","Total_Sales"]

for col in numeric_columns:

  Q1 = df[col].quantile(0.25)
  Q2 = df[col].quantile(0.75)

  IQR = Q2 -Q1

  lower = Q1 - 1.5 * IQR
  upper = Q2 + 1.5 * IQR

  Outliers = df[(df[col]  < lower) | (df[col] > upper) ]

  print(col, "Outliers:", len(Outliers))
```

```python
df["Calculated_Sales"] = (
    df["Quantity"] * df["Unit_Price"]
)

print(
    df[df["Total_Sales"] != df["Calculated_Sales"]]
)
```

```python
# Fill missing Age with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing City with Unknown
df["City"] = df["City"].fillna("Unknown")

print(df.isnull().sum())
```

```python
#Standardize date and text formate

df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],errors="coerce"
)

text_columns = ["Gender","City","Product","Category"]

for col in text_columns:
  df[col] = df[col].str.strip().str.title()

  print(df.dtypes)
```

```python
# Identify repeated Order IDs
df["Duplicate_Order_ID"] = df["Order_ID"].duplicated(
    keep=False
)

print(
    df[df["Duplicate_Order_ID"]]
)
```

```python
#Feature Engineering

df["Order_Year"] = df["Order_Date"].dt.year

df["Order_Month"] = df["Order_Date"].dt.month

df["Month_Name"] =  df["Order_Date"].dt.month_name()

df["Quarter"] = df["Order_Date"].dt.quarter

# Create Age Groups

df["Age_Group"] = pd.cut(df["Age"], bins=[0,25,35,45,55,100], labels = ["18-25","26-35","36-45","46-55","Above 55"], include_lowest=True)

print(df.head())
```

```python
Q1 = df["Total_Sales"].quantile(0.25)
Q3 = df["Total_Sales"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df["Sales_Outlier"] = (
    (df["Total_Sales"] < lower) |
    (df["Total_Sales"] > upper)
)

print(df["Sales_Outlier"].value_counts())
```

```python
# Genrating Final Analysis Ready DataSet

df.to_csv("ApexPlanet_Cleaned_Dataset.csv", index=False)

df.to_excel("ApexPlanet_Cleaned_Dataset.xlsx", index=False)

print("Dataset Cleaned And Saved Successfully....")
```
